```python
# **Q2.** Write a function is_prime(n) that returns True if n is a prime number, otherwise False.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in [1, 2, 15, 17, 29, 30]:
    print(f"is_prime({num}) = {is_prime(num)}")
```

```python
# **Q3.** Write a program to print the Fibonacci sequence up to n terms using a loop.
def fibonacci(n):
    seq = []
    a, b = 0, 1
    for _ in range(n):
        seq.append(a)
        a, b = b, a + b
    return seq

n = 10
print(f"First {n} Fibonacci numbers: {fibonacci(n)}")
```

```python
# 
def remove_duplicates(nums):
    seen = set()
    result = []
    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result

sample = [4, 2, 7, 2, 4, 9, 1, 7, 3]
print("Original:", sample)
print("Duplicates removed:", remove_duplicates(sample))
```

```python
# **Q5.** Write a program that counts the number of vowels, consonants, digits, and spaces in a given string.
def count_chars(text):
    vowels = consonants = digits = spaces = 0
    for ch in text:
        if ch.isalpha():
            if ch.lower() in "aeiou":
                vowels += 1
            else:
                consonants += 1
        elif ch.isdigit():
            digits += 1
        elif ch.isspace():
            spaces += 1
    return vowels, consonants, digits, spaces

s = "Hello World 123, Machine Learning Lab 1"
v, c, d, sp = count_chars(s)
print(f"String: {s!r}")
print(f"Vowels: {v}, Consonants: {c}, Digits: {d}, Spaces: {sp}")
```

```python
# **Q13.** Given a NumPy array of 25 random integers between 1 and 100, find the sum, mean, and standard deviation.
import numpy as np

np.random.seed(42)
arr = np.random.randint(1, 101, size=25)
print("Array:\n", arr)
print(f"\nSum: {arr.sum()}")
print(f"Mean: {arr.mean():.2f}")
print(f"Standard Deviation: {arr.std():.2f}")
```

```python
# **Q14.** Given a 2D NumPy array of shape (4,4), extract the diagonal elements using np.diag() and compute their sum.
import numpy as np

arr = np.arange(1, 17).reshape(4, 4)
print("4x4 array:\n", arr)

diagonal = np.diag(arr)
print("\nDiagonal elements:", diagonal)
print("Sum of diagonal:", diagonal.sum())
```

```python
# **Q15.** Create two NumPy arrays of shape (3,3) and demonstrate the difference 
# between element-wise multiplication (*) and matrix multiplication (@).
import numpy as np

A = np.arange(1, 10).reshape(3, 3)
B = np.arange(9, 0, -1).reshape(3, 3)

print("A:\n", A)
print("\nB:\n", B)

elementwise = A * B
matmul = A @ B

print("\nElement-wise multiplication (A * B):\n", elementwise)
print("\nMatrix multiplication (A @ B):\n", matmul)
```

## Section C — Matplotlib & Pandas (Q26–Q30)

```python
# **Q26.** Given a NumPy array of 30 days of temperature readings, plot a line chart with the 
# day number on the x-axis and temperature on the y-axis, including a title and axis labels.
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
days = np.arange(1, 31)
temps = 30 + np.random.randn(30) * 3

plt.figure(figsize=(8, 4))
plt.plot(days, temps, marker='o', color='tab:blue')
plt.title("Daily Temperature Readings (30 Days)")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# **Q27.** Plot y = sin(x) and y = cos(x) on the same figure for x from 0 to 2π using np.linspace(); 
# include a legend distinguishing the two curves.
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y_sin, label="sin(x)", color="tab:orange")
plt.plot(x, y_cos, label="cos(x)", color="tab:green")
plt.title("y = sin(x) and y = cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# **Q28.** Create a bar chart comparing the average marks of 4 subjects, 
# with a distinct color for the highest-scoring subject.
import numpy as np
import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "English", "CS"]
avg_marks = [78, 85, 72, 90]

max_idx = int(np.argmax(avg_marks))
colors = ["tab:blue"] * len(subjects)
colors[max_idx] = "tab:red"

plt.figure(figsize=(7, 4))
plt.bar(subjects, avg_marks, color=colors)
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
for i, v in enumerate(avg_marks):
    plt.text(i, v + 0.5, str(v), ha="center")
plt.tight_layout()
plt.show()
```

```python
# **Q29.** Generate a histogram of 500 normally distributed random numbers and overlay a 
# vertical line at the mean using plt.axvline().
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(7)
data = np.random.normal(loc=0, scale=1, size=500)
mean_val = data.mean()

plt.figure(figsize=(7, 4))
plt.hist(data, bins=30, color="tab:purple", alpha=0.7, edgecolor="black")
plt.axvline(mean_val, color="red", linestyle="--", linewidth=2, label=f"Mean = {mean_val:.2f}")
plt.title("Histogram of 500 Normally Distributed Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# **Q30.** Create a 2x2 grid of subplots showing four different chart types 
# (line, bar, scatter, histogram) from the same dataset.
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(3)
x = np.arange(1, 21)
y = np.random.randint(10, 100, size=20)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, y, color="tab:blue")
axes[0, 0].set_title("Line Chart")

axes[0, 1].bar(x, y, color="tab:green")
axes[0, 1].set_title("Bar Chart")

axes[1, 0].scatter(x, y, color="tab:orange")
axes[1, 0].set_title("Scatter Plot")

axes[1, 1].hist(y, bins=8, color="tab:red", edgecolor="black")
axes[1, 1].set_title("Histogram")

fig.suptitle("Same Dataset — Four Chart Types")
plt.tight_layout()
plt.show()
```
